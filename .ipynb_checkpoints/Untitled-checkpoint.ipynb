Ручками записываем из программы

```python
predictions = [239041, 578610, 368921, 756052, 239798, 360239, 949323, 342317, 373975, 1260076, 3348, 19000, 12038, 2462, 1285, 3051, 1380, 1271, 3462, 2794]
```

```python
df['predicted_price'] = predictions
```

```python
df['name'] = 'dinislam'
```

```python
df['group'] = '23p-4'
```

```python
df.head(20)
```

```python
x = df['price']
```

```python
y = df['predicted_price']
```

```python
import numpy as np
```

```python
def mae(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    return np.mean(np.abs(x - y))
```

```python
def rmse(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    return np.sqrt(np.mean((x - y) ** 2))
```

```python
def r2(x, y):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    ss_res = np.sum((x - y) ** 2)
    ss_tot = np.sum((x - np.mean(x)) ** 2)
    return 1 - ss_res / ss_tot
```

```python
print(f"MAE: {mae(x,y):.2f}")
print(f"RMSE: {rmse(x,y):.2f}")
print(f"R2: {r2(x,y):.2f}")
```

Это абсолютная в лирах и там есть большое расхождение от 1000 до 250000 +- а также смешана аренда с продажей

```python
x_log = np.log1p(df['price'].values)
y_log = np.log1p(df['predicted_price'].values)
```

```python
print(f"MAE: {mae(x_log,y_log):.2f}")
print(f"RMSE: {rmse(x_log,y_log):.2f}")
print(f"R2: {r2(x_log,y_log):.2f}")
```

Уже в лог-шкале они более точные и подходят для оценивания модели

```python
df['mae'] = f"{mae(x_log,y_log):.2f}"
df['rmse'] = f"{rmse(x_log,y_log):.2f}"
df['r2'] = f"{r2(x_log,y_log):.2f}"
```

```python
for lt in [1, 2]:
    sub = df[df['listing_type'] == lt]
    x_s, y_s = sub['price'].values, sub['predicted_price'].values
    x_log = np.log1p(x_s)
    y_log = np.log1p(y_s)
    print(f"тип{lt}: MAE: {mae(x_s,y_s):.2f}, RMSE: {rmse(x_s,y_s):.2f}, R2: {r2(x_s,y_s):.4f}")
    print(f"LOG: тип{lt}: MAE: {mae(x_log,y_log):.2f}, RMSE: {rmse(x_log,y_log):.2f}, R2: {r2(x_log,y_log):.4f}")
```

Из итого модель по продажам в лирах попадает 100%, но не точно, а лишь в диапазон. Он хорошо распознаёт разницу в продаже и аренде. Но уже внутри типо он слаб.
